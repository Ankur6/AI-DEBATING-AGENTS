````python
# app/agents/ingestion.py
!pip install langchain-openai
from langchain_openai import ChatOpenAI
from langchain_core.prompts import ChatPromptTemplate
from pydantic import BaseModel, Field
from typing import List

class DecomposedClaims(BaseModel):
    sub_claims: List[str] = Field(description="List of atomic, verifiable facts extracted from the input.")

class IngestionAgent:
    def __init__(self):
        self.llm = ChatOpenAI(model="gpt-4o", temperature=0)
        self.parser = self.llm.with_structured_output(DecomposedClaims)

    def parse(self, user_input: str) -> List[str]:
        prompt = ChatPromptTemplate.from_messages([
            ("system", "You are a forensic linguistic analyst. Break the following complex claim into individual, atomic statements that can be fact-checked independently."),
            ("user", "{input}")
        ])
        chain = prompt | self.parser
        result = chain.invoke({"input": user_input})
        return result.sub_claims
````

### Subtask:

**Reasoning**:
The previous attempt to run the FastAPI application failed because `app/main.py` was not properly created or found. Now that `app/main.py` has been correctly written, I will stop any potentially lingering `uvicorn` processes, and then restart the FastAPI application in the background using `uvicorn`, redirecting output to `nohup.out` and adding a delay to allow for startup.



````python
import subprocess
import time
import os

# 1. Stop any existing uvicorn process on port 8000
print("Attempting to stop any existing uvicorn process on port 8000...")
# Use lsof to find PIDs listening on port 8000 and kill them
# || true is added to prevent the script from failing if no process is found
stop_command = "lsof -t -i :8000 | xargs kill -9 || true"
subprocess.run(stop_command, shell=True, capture_output=True, text=True)
print("Checked for and stopped previous processes.")

# 2. Restart the FastAPI application in the background
# Construct the command as a single string to be executed by the shell
command = "nohup uvicorn app.main:app --host 0.0.0.0 --port 8000 > nohup.out 2>&1 &"

# Run the FastAPI application in the background using shell=True
# The nohup command ensures the process continues even if the terminal is closed.
# The '&' sends the process to the background.
process = subprocess.Popen(command, shell=True)

# Optionally, print a message indicating the process was launched
print(f"FastAPI application launched in background. Check 'nohup.out' for logs.")
print(f"Process PID (from Python Popen): {process.pid}")

# 3. Give the server a moment to start up
time.sleep(5)

# Verify if the process is still running and check nohup.out for Uvicorn's actual PID
print("Checking nohup.out for server status...")
````
